## Theory
**Divide and Conquer** is an algorithmic paradigm that solves a complex problem by breaking it down into smaller subproblems, solving the subproblems independently (often recursively), and then combining their solutions to solve the original problem.

The Divide and Conquer strategy consists of three main steps:
1. **Divide:** Partition the original problem into two or more smaller, independent subproblems of the same type.
2. **Conquer:** Solve the subproblems recursively. If subproblems are small enough (base case), solve them directly.
3. **Combine:** Merge the solutions of the subproblems into the solution for the original problem.

### Algorithms Overview:
- **Binary Search:** Efficiently locates a target value within a **sorted array** by halving the search space recursively or iteratively. Recurrence: $T(n) = T(n/2) + O(1) \implies O(\log n)$.
- **Minimum and Maximum (Iterative vs D&C):** Iterative approach scans linearly performing $2(n-1)$ comparisons. D&C approach divides array into halves, finds min-max recursively, and combines results in $\frac{3n}{2} - 2$ comparisons.
- **Merge Sort:** Divide the array into two equal halves, recursively sort both halves, and merge the sorted halves. Recurrence: $T(n) = 2T(n/2) + O(n) \implies O(n \log n)$.
- **Heap Sort:** Uses a Binary Heap data structure (Max-Heap). First transforms the array into a max heap, then repeatedly extracts the max element and restores heap property. Time Complexity: $O(n \log n)$.
- **Quick Sort:** Selects a 'pivot' element, partitions the array such that elements smaller than pivot are on the left and larger on the right, then recursively sorts partitions. Recurrence: $T(n) = T(k) + T(n-k-1) + O(n) \implies O(n \log n)$ average, $O(n^2)$ worst case.


## 1. WAP to implement Binary Search

### Algorithm
1. **Start**
2. Input a sorted array `arr` and the target value `target`.
3. Set `low = 0` and `high = len(arr) - 1`.
4. Repeat while `low <= high`:
   - Calculate `mid = (low + high) // 2`.
   - If `arr[mid] == target`, return `mid` (element found).
   - If `target < arr[mid]`, update `high = mid - 1` (search left half).
   - Else, update `low = mid + 1` (search right half).
5. If loop terminates without match, return `-1` (element not found).
6. **Stop**

```python
def binary_search(arr, target):
    low = 0
    high = len(arr) - 1

    while low <= high:
        mid = (low + high) // 2

        if arr[mid] == target:
            return mid
        elif target < arr[mid]:
            high = mid - 1
        else:
            low = mid + 1

    return -1

# Demonstration with a sorted array
arr = [2, 5, 8, 12, 16, 23, 38, 56, 72, 91]
target = 23

print("Sorted Array:", arr)
print("Target Element:", target)

result = binary_search(arr, target)

if result != -1:
    print(f"Element {target} found at index {result}")
else:
    print("Element not found")
```

## 2. WAP to find min and max of an array using an iterative process

### Algorithm
1. **Start**
2. Input the array `arr` of size `n`.
3. Initialize `minimum = arr[0]` and `maximum = arr[0]`.
4. Loop through each element `num` in `arr` from index 1 to `n-1`:
   - If `num < minimum`, set `minimum = num`.
   - If `num > maximum`, set `maximum = num`.
5. Display `minimum` and `maximum`.
6. **Stop**

```python
arr = [82, 13, 214, 532, 2, 3, 45, 67, 89]

minimum = arr[0]
maximum = arr[0]

for num in arr[1:]:
    if num < minimum:
        minimum = num
    if num > maximum:
        maximum = num

print("Array:", arr)
print("Minimum element:", minimum)
print("Maximum element:", maximum)
```

## 3. WAP to find min and max of an array using a divide and conquer algorithm

### Algorithm
1. **Start**
2. Input array `arr` and range bounds `low` and `high`.
3. Define recursive function `find_min_max(arr, low, high)`:
   - **Base Case 1:** If `low == high`, return `(arr[low], arr[low])`.
   - **Base Case 2:** If `high == low + 1`, compare `arr[low]` and `arr[high]`, return `(min, max)`.
   - **Recursive Case:**
     - Calculate `mid = (low + high) // 2`.
     - `(left_min, left_max) = find_min_max(arr, low, mid)`
     - `(right_min, right_max) = find_min_max(arr, mid + 1, high)`
     - Return `(min(left_min, right_min), max(left_max, right_max))`.
4. Call `find_min_max(arr, 0, len(arr) - 1)`.
5. Display `minimum` and `maximum`.
6. **Stop**

```python
def find_min_max(arr, low, high):
    # Base Case: Single element
    if low == high:
        return arr[low], arr[low]

    # Base Case: Two elements
    if high == low + 1:
        if arr[low] < arr[high]:
            return arr[low], arr[high]
        else:
            return arr[high], arr[low]

    # Divide step
    mid = (low + high) // 2

    # Conquer step
    left_min, left_max = find_min_max(arr, low, mid)
    right_min, right_max = find_min_max(arr, mid + 1, high)

    # Combine step
    minimum = min(left_min, right_min)
    maximum = max(left_max, right_max)

    return minimum, maximum

arr = [15, 8, 23, 4, 19, 2, 31, 10]
minimum, maximum = find_min_max(arr, 0, len(arr) - 1)

print("Array:", arr)
print("Minimum element:", minimum)
print("Maximum element:", maximum)
```

## 4. WAP to implement Merge Sort

### Algorithm
1. **Start**
2. Input the unsorted array `arr`.
3. Define `merge_sort(arr, l, r)`:
   - If `l < r`:
     - Calculate midpoint `m = (l + r) // 2`.
     - Call `merge_sort(arr, l, m)` to sort left subarray.
     - Call `merge_sort(arr, m + 1, r)` to sort right subarray.
     - Call `merge(arr, l, m, r)` to merge the two sorted subarrays.
4. Define `merge(arr, l, m, r)`:
   - Create temporary arrays for left and right subarrays.
   - Compare elements of left and right temporary arrays and copy smaller element into `arr`.
   - Copy any remaining elements from temporary arrays into `arr`.
5. Display original and sorted array.
6. **Stop**

```python
def merge(arr, l, m, r):
    n1 = m - l + 1
    n2 = r - m

    L = arr[l:m + 1]
    R = arr[m + 1:r + 1]

    i = 0
    j = 0
    k = l

    while i < n1 and j < n2:
        if L[i] <= R[j]:
            arr[k] = L[i]
            i += 1
        else:
            arr[k] = R[j]
            j += 1
        k += 1

    while i < n1:
        arr[k] = L[i]
        i += 1
        k += 1

    while j < n2:
        arr[k] = R[j]
        j += 1
        k += 1

def merge_sort(arr, l, r):
    if l < r:
        m = (l + r) // 2
        merge_sort(arr, l, m)
        merge_sort(arr, m + 1, r)
        merge(arr, l, m, r)

arr = [38, 27, 43, 3, 9, 82, 10]
print("Original Array:", arr)

merge_sort(arr, 0, len(arr) - 1)
print("Sorted Array (Merge Sort):", arr)
```

## 5. WAP to implement Heap Sort

### Algorithm
1. **Start**
2. Input the unsorted array `arr` of size `n`.
3. Define `heapify(arr, n, i)`:
   - Set `largest = i`, `left = 2*i + 1`, `right = 2*i + 2`.
   - If `left < n` and `arr[left] > arr[largest]`, update `largest = left`.
   - If `right < n` and `arr[right] > arr[largest]`, update `largest = right`.
   - If `largest != i`, swap `arr[i]` and `arr[largest]`, then recursively call `heapify(arr, n, largest)`.
4. Define `heap_sort(arr)`:
   - **Build Max-Heap:** Loop `i` from `n//2 - 1` down to 0, call `heapify(arr, n, i)`.
   - **Extract Elements:** Loop `i` from `n-1` down to 1:
     - Swap `arr[0]` (root max element) with `arr[i]`.
     - Call `heapify(arr, i, 0)` on the reduced heap.
5. Display original and sorted array.
6. **Stop**

```python
def heapify(arr, n, i):
    largest = i
    left = 2 * i + 1
    right = 2 * i + 2

    if left < n and arr[left] > arr[largest]:
        largest = left

    if right < n and arr[right] > arr[largest]:
        largest = right

    if largest != i:
        arr[i], arr[largest] = arr[largest], arr[i]
        heapify(arr, n, largest)

def heap_sort(arr):
    n = len(arr)

    # Build max-heap
    for i in range(n // 2 - 1, -1, -1):
        heapify(arr, n, i)

    # Extract elements one by one from heap
    for i in range(n - 1, 0, -1):
        arr[i], arr[0] = arr[0], arr[i]
        heapify(arr, i, 0)

arr = [12, 11, 13, 5, 6, 7]
print("Original Array:", arr)

heap_sort(arr)
print("Sorted Array (Heap Sort):", arr)
```

## 6. WAP to implement Quick Sort

### Algorithm
1. **Start**
2. Input the unsorted array `arr`.
3. Define `partition(arr, low, high)`:
   - Choose `pivot = arr[high]`.
   - Initialize index `i = low - 1`.
   - Loop `j` from `low` to `high - 1`:
     - If `arr[j] <= pivot`, increment `i` and swap `arr[i]` with `arr[j]`.
   - Swap `arr[i + 1]` with `arr[high]`.
   - Return partition index `i + 1`.
4. Define `quick_sort(arr, low, high)`:
   - If `low < high`:
     - Compute `pi = partition(arr, low, high)`.
     - Recursively call `quick_sort(arr, low, pi - 1)`.
     - Recursively call `quick_sort(arr, pi + 1, high)`.
5. Display original and sorted array.
6. **Stop**

```python
def partition(arr, low, high):
    pivot = arr[high]
    i = low - 1

    for j in range(low, high):
        if arr[j] <= pivot:
            i += 1
            arr[i], arr[j] = arr[j], arr[i]

    arr[i + 1], arr[high] = arr[high], arr[i + 1]
    return i + 1

def quick_sort(arr, low, high):
    if low < high:
        pi = partition(arr, low, high)
        quick_sort(arr, low, pi - 1)
        quick_sort(arr, pi + 1, high)

arr = [10, 7, 8, 9, 1, 5]
print("Original Array:", arr)

quick_sort(arr, 0, len(arr) - 1)
print("Sorted Array (Quick Sort):", arr)
```

# Analysis of the algorithms

| Algorithm | Time Complexity (Best) | Time Complexity (Average) | Time Complexity (Worst) | Space Complexity | Notes |

| *Merge Sort* | $\Omega(n \log n)$ | $\Theta(n \log n)$ | $O(n \log n)$ | $O(n)$ | Stable sorting algorithm, extra memory needed for merging. |

| *Quick Sort* | $\Omega(n \log n)$ | $\Theta(n \log n)$ | $O(n^2)$ | $O(\log n)$ stack | In-place partition sort, highly efficient in practice. |

# Discussion

In this experiment, various Divide and Conquer algorithms were implemented and evaluated in Python alongside an iterative algorithm for comparison:

1. **Binary Search** demonstrates how dividing the problem domain in half at each step reduces logarithmic search time $O(\log n)$ compared to linear search $O(n)$.
2. **Min-Max Determination:** While both iterative and Divide and Conquer methods achieve $O(n)$ time complexity, the Divide and Conquer approach optimizes element comparison operations to $\frac{3n}{2} - 2$, which is significantly lower than the $2n - 2$ comparisons required by the simple iterative loop.
3. **Sorting Algorithms Comparison:**
   - **Merge Sort** guarantees $O(n \log n)$ time complexity across all cases (best, average, worst) due to its balanced splitting strategy. However, it requires $O(n)$ additional space for auxiliary arrays during the merge operation.
   - **Heap Sort** also achieves $O(n \log n)$ time complexity in all cases while maintaining $O(1)$ space complexity by performing in-place restructuring of a binary max-heap.
   - **Quick Sort** exhibits superior practical performance due to low constant factors and in-place partitioning ($O(\log n)$ auxiliary recursion stack). Its worst-case time complexity $O(n^2)$ occurs when bad pivots (e.g. already sorted array with extreme element selection) are chosen, but can be mitigated using randomized pivot selection or median-of-three rules.
