# Theory

Iterative and numerical algorithms are fundamental techniques in computer science and mathematics used to solve computational problems efficiently. An iterative algorithm repeatedly executes a set of instructions until a specified condition is met, making it more memory-efficient than recursive approaches. Numerical algorithms involve mathematical computations such as modular arithmetic, prime number testing, searching, and sorting, and are widely used in scientific computing, data processing, and cryptography.

The algorithms implemented in this experiment are described below:

Euclid's Algorithm is one of the oldest and most efficient methods for finding the Greatest Common Divisor (GCD) of two integers. It repeatedly replaces the larger number with the remainder of dividing it by the smaller number until the remainder becomes zero. The last non-zero remainder is the GCD.

Fibonacci Series generates a sequence in which each number is the sum of the two preceding numbers, beginning with 0 and 1. It demonstrates the concept of iteration and sequence generation.

Sequential Search (Linear Search) is the simplest searching algorithm. It checks each element of a list one by one until the desired element is found or the end of the list is reached. Although simple to implement, it is less efficient for large datasets.

Bubble Sort is a comparison-based sorting algorithm that repeatedly compares adjacent elements and swaps them if they are in the wrong order. After each pass, the largest unsorted element moves to its correct position.

Insertion Sort builds a sorted list one element at a time by inserting each new element into its proper position among the previously sorted elements. It performs efficiently for small or nearly sorted datasets.

Selection Sort repeatedly selects the smallest element from the unsorted portion of the list and places it at the beginning. It is easy to understand and implement but requires more comparisons than some advanced sorting algorithms.

Chinese Remainder Theorem (CRT) is a mathematical theorem used to solve systems of simultaneous congruences with pairwise coprime moduli. It has important applications in cryptography, computer security, and modular arithmetic.

Fermat's Little Theorem states that if p is a prime number and a is an integer not divisible by p, then a^(p−1) ≡ 1 (mod p). It is widely used in modular exponentiation, cryptography, and preliminary primality testing.

Miller-Rabin Primality Test is a probabilistic algorithm used to determine whether a number is prime. It performs multiple rounds of testing to identify composite numbers with high accuracy and is commonly used in modern cryptographic systems due to its speed and reliability.

These algorithms provide the foundation for many advanced applications in algorithm design, data structures, cryptography, computer security, database management, scientific computing, and optimization techniques. Understanding their implementation helps in developing efficient software solutions and strengthens problem-solving skills in programming.


## 1. WAP to implement GCD using the Euclid algorithm

### Algorithm 
* Start.
* Input two integers a and b.
* Repeat while b is not equal to 0.
* Store b in a temporary variable.
* Replace b with a % b.
* Replace a with the temporary value.
* Display a as the GCD.
* Stop.

```python
def gcd(a, b):
    while b != 0:
        a, b = b, a % b
    return a

a = int(input("Enter first number: "))
b = int(input("Enter second number: "))

print("GCD =", gcd(a, b))
```

* Repeat n times:

## 3. WAP to implement Sequential Search 

### Algorithm 
* Start.
* Input the list of elements.
* Input the element to be searched.
* Compare the search element with each element in the list one by one.
* If the element is found:
* Display its position.
* Stop searching.
* If the element is not found after checking all elements:
* Display "Element not found."
* Stop.

```python
arr = [12, 45, 78, 23, 56, 90]

key = int(input("Enter element to search: "))

found = False

for i in range(len(arr)):
    if arr[i] == key:
        print("Element found at index", i)
        found = True
        break

if not found:
    print("Element not found")
```

## 4.WAP to implement Bubble Sort

### Algorithm
* Start.
* Input the array.
* Compare each adjacent pair of elements.
* If the first element is greater than the second, swap them.
* Repeat the process for all elements.
* Continue passes until the array is sorted.
* Display the sorted array.
* Stop.

```python
arr = [64, 34, 25, 12, 22, 11, 90]

n = len(arr)

for i in range(n):
    for j in range(0, n-i-1):
        if arr[j] > arr[j+1]:
            arr[j], arr[j+1] = arr[j+1], arr[j]

print("Sorted Array:")
print(arr)
```

## 5.WAP to implement Insertion Sort

### Algorithm
* Start.
* Input the array.
* Assume the first element is already sorted.
* Pick the next element as the key.
* Compare the key with previous elements.
* Shift larger elements one position to the right.
* Insert the key into its correct position.
* Repeat until all elements are sorted.
* Display the sorted array.
* Stop.

```python
arr = [12, 11, 13, 5, 6]

for i in range(1, len(arr)):
    key = arr[i]
    j = i - 1

    while j >= 0 and arr[j] > key:
        arr[j + 1] = arr[j]
        j -= 1

    arr[j + 1] = key

print(arr)
```

## 6.WAP to implement Selection Sort

### Algorithm 
* Start.
* Input the array.
* Find the smallest element in the unsorted portion.
* Swap it with the first unsorted element.
* Move the sorted boundary one position forward.
* Repeat until all elements are sorted.
* Display the sorted array.
* Stop.

```python
arr = [64, 25, 12, 22, 11]

for i in range(len(arr)):
    minimum = i

    for j in range(i+1, len(arr)):
        if arr[j] < arr[minimum]:
            minimum = j

    arr[i], arr[minimum] = arr[minimum], arr[i]

print(arr)
```

## 7.WAP to implement the Chinese Remainder Theorem

### Algorithm 
* Start.
* Input the moduli and their corresponding remainders.
* Calculate the product of all moduli.
* For each modulus:
* Calculate the partial product.
* Find its modular inverse.
* Multiply the remainder, modular inverse, and partial product.
* Add the result to the final sum.
* Compute the final answer by taking the modulus with the product.
* Display the solution.
* Stop.

```python
from functools import reduce

num = [3, 5, 7]
rem = [2, 3, 2]

product = reduce(lambda x, y: x * y, num)

result = 0

for n, r in zip(num, rem):
    pp = product // n

    inv = 1
    while (pp * inv) % n != 1:
        inv += 1

    result += r * inv * pp

print("Solution =", result % product)
```

## 8.WAP to implement Fermat's Theorem

### Algorithm 
* Start.
* Input an integer a and a prime number p.
* Calculate a^(p−1) mod p.
* If the result is 1, the theorem is satisfied.
* Display the result.
* Stop.

```python
def power(a, b, p):
    result = 1
    a = a % p

    while b > 0:
        if b % 2 == 1:
            result = (result * a) % p

        b = b // 2
        a = (a * a) % p

    return result

p = int(input("Enter prime number: "))
a = int(input("Enter integer: "))

print(power(a, p-1, p))
```

## 9.WAP to implement Millar - Rabin Test

### Algorithm 
* Start.
* Input the integer n.
* If n is less than or equal to 1, declare it composite.
* Express n−1 as 2^r × d, where d is odd.
* Repeat the test for a fixed number of iterations:
* Select a random integer a.
* Compute x = a^d mod n.
* If x is 1 or n−1, continue.
* Otherwise, repeatedly square x modulo n.
* If x never becomes n−1, declare n composite.
* If all iterations pass, declare n probably prime.
* Stop.

```python
import random

def miller_rabin(n, k=5):
    if n <= 1:
        return False
    if n <= 3:
        return True

    d = n - 1
    r = 0

    while d % 2 == 0:
        d //= 2
        r += 1

    for _ in range(k):
        a = random.randint(2, n - 2)
        x = pow(a, d, n)

        if x == 1 or x == n - 1:
            continue

        for _ in range(r - 1):
            x = pow(x, 2, n)

            if x == n - 1:
                break
        else:
            return False

    return True

n = int(input("Enter number: "))

if miller_rabin(n):
    print("Probably Prime")
else:
    print("Composite")
```
